```python
import random
jobs = [5, 3, 8, 6, 4, 7]
num_machines = 3
num_wolves = 10
max_iterations = 50
def fitness(wolf):
    machine_load = [0] * num_machines

    for i in range(len(jobs)):
        machine = int(round(wolf[i]))
        machine = max(0, min(machine, num_machines - 1))

        machine_load[machine] += jobs[i]

    return max(machine_load)
wolves = [
    [random.uniform(0, num_machines - 1)
     for _ in range(len(jobs))]
    for _ in range(num_wolves)
]

for iteration in range(max_iterations):

    wolves.sort(key=fitness)
    alpha = wolves[0][:]
    beta = wolves[1][:]
    delta = wolves[2][:]
    a = 2 - (2 * iteration / max_iterations)
    for i in range(num_wolves):

        for j in range(len(jobs)):
            r1 = random.random()
            r2 = random.random()

            A1 = 2 * a * r1 - a
            C1 = 2 * r2
            D_alpha = abs(C1 * alpha[j] - wolves[i][j])
            X1 = alpha[j] - A1 * D_alpha
            r1 = random.random()
            r2 = random.random()
            A2 = 2 * a * r1 - a
            C2 = 2 * r2
            D_beta = abs(C2 * beta[j] - wolves[i][j])
            X2 = beta[j] - A2 * D_beta
            r1 = random.random()
            r2 = random.random()
            A3 = 2 * a * r1 - a
            C3 = 2 * r2
            D_delta = abs(C3 * delta[j] - wolves[i][j])
            X3 = delta[j] - A3 * D_delta
            wolves[i][j] = (X1 + X2 + X3) / 3
            wolves[i][j] = max(
                0, min(wolves[i][j], num_machines - 1)
            )

wolves.sort(key=fitness)

best_wolf = wolves[0]
best_fitness = fitness(best_wolf)
machine_load = [0] * num_machines

print("Job Scheduling using GWO")
print("------------------------")
for i in range(len(jobs)):
    machine = int(round(best_wolf[i]))
    machine = max(0, min(machine, num_machines - 1))
    machine_load[machine] += jobs[i]
    print("Job", i + 1, "-> Machine", machine + 1)
print("\nMachine Loads:")
for i in range(num_machines):
    print("Machine", i + 1, ":", machine_load[i])
print("\nMinimum Makespan:", best_fitness)
```
