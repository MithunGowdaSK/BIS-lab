```python


import numpy as np

def fitness(x):
    return x ** 2

def cuckoo_search():

    n = 10
    Pa = 0.25
    MaxIter = 100

    nests = np.random.uniform(-10, 10, n)

    scores = np.array([fitness(x) for x in nests])

    best_index = np.argmin(scores)
    best = nests[best_index]

    for t in range(MaxIter):

        for i in range(n):

            step = np.random.normal(0, 1)
            new_solution = nests[i] + step

            new_fitness = fitness(new_solution)

            j = np.random.randint(n)

            if new_fitness < scores[j]:
                nests[j] = new_solution
                scores[j] = new_fitness

        for i in range(n):
            if np.random.rand() < Pa:
                nests[i] = np.random.uniform(-10, 10)
                scores[i] = fitness(nests[i])

        best_index = np.argmin(scores)

        if scores[best_index] < fitness(best):
            best = nests[best_index]

    print("Best solution (x):", best)
    print("Minimum fitness value:", fitness(best))

cuckoo_search()
```
